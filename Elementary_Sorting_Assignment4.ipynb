## 1. Comparable API (`__lt__`)

## Idée

Les algorithmes de tri ne dépendent pas du type des données.

Ils utilisent seulement :

```python
a[i] < a[j]
```

Python appelle automatiquement :

```python
a[i].__lt__(a[j])
```

---

**Exemple**

```python
class Student:

    def __lt__(self, other):

        if self.grade != other.grade:
            return self.grade < other.grade

        if self.surname != other.surname:
            return self.surname < other.surname

        return self.firstname < other.firstname
```

---

**Important**

Le comparateur doit définir :

**un ordre total cohérent.**

Donc impossible que :

```text
a < b = True
b < a = True
```

---

## 2. Selection Sort

## Idée

À chaque étape :

* chercher le minimum,
* l’échanger avec la position courante.

---

**Implémentation**

```python
def selectionSort(a):

    N = len(a)

    for i in range(N):

        minIndex = i

        for j in range(i+1, N):

            if a[j] < a[minIndex]:
                minIndex = j

        a[i], a[minIndex] = a[minIndex], a[i]
```

---

**Complexité**

## Comparaisons

Toujours :

$\frac{N(N-1)}{2}\sim \frac12 N^2$

---

## Échanges

Seulement :

$N$

---

**Temps**

| cas     | complexité |
| ------- | ---------- |
| best    | (N^2)      |
| average | (N^2)      |
| worst   | (N^2)      |

---

**Propriété importante**

Insensible à l’ordre initial.

Même tableau déjà trié :

**toujours quadratique.**

---

**Avantage**

Très peu de mouvements mémoire.

---

## 3. Insertion Sort

## Idée

Construire progressivement une partie triée à gauche.

Chaque nouvel élément :

* remonte vers la gauche.

---

**Implémentation**

```python
def insertionSort(a):

    for i in range(1, len(a)):

        j = i

        while j > 0 and a[j] < a[j-1]:

            a[j], a[j-1] = a[j-1], a[j]

            j -= 1
```

---

**Complexités**

| cas     | complexité |
| ------- | ---------- |
| best    | (N)        |
| average | (N^2)      |
| worst   | (N^2)      |

---

**Best case**

Tableau déjà trié :

```text
A E E L M O P R S T X
```

Seulement :

$N-1$

comparaisons.

Aucun échange.

---

**Worst case**

Tableau inversé :

```text
X T S R P O M L E E A
```

Environ :

$\frac12 N^2$

comparaisons et échanges.

---

**Average case**

En moyenne :

$\frac14 N^2$

comparaisons et échanges.

---

## 4. Inversions

## Définition

Une inversion =

**une paire mal ordonnée.**

Exemple :

```text
A E E L M O T R X P S
```

Inversions :

* T-R
* T-P
* T-S
* R-P
* X-P
* X-S

---

**Propriété**

Insertion sort :

**nombre d’échanges = nombre d’inversions.**

---

**Très important**

Si le tableau est presque trié :

**insertion sort devient très rapide.**

---

**Partially sorted arrays**

Tableau partiellement trié :

$\text{inversions} \le cN$

Alors insertion sort devient :

$O(N)$

---

## 5. Improvements of Insertion Sort

---

**Half exchanges (shifting)**

Au lieu de swap successifs :

* décaler les éléments,
* insérer directement.

---

**Binary insertion sort**

Utiliser binary search pour trouver la position.

---

**Gain**

Comparaisons :

$N\log N$

---

**MAIS**

Les décalages restent :

$N^2$

Donc :

**complexité globale toujours quadratique.**

## 6. Shellsort

## Idée principale

Déplacer les éléments de plusieurs positions à la fois.

---

**h-sorted**

Un tableau est h-sorted si :

$a[i]\le a[i+h]$

---

**Principe**

Faire plusieurs insertion sorts :

* avec grands gaps,
* puis petits gaps,
* puis finalement :

```text
h = 1
```

---

**Suite 3x+1**

```text
1, 4, 13, 40, 121, ...
```

Calcul :

```python
h = 3*h + 1
```

---

**Implémentation**

```python
def shellSort(a):

    N = len(a)

    h = 1

    while h < N//3:
        h = 3*h + 1

    while h >= 1:

        for i in range(h, N):

            j = i

            while j >= h and a[j] < a[j-h]:

                a[j], a[j-h] = a[j-h], a[j]

                j -= h

        h = h//3
```

---

**Complexités Shellsort (3x+1)**

| cas     | complexité                     |
| ------- | ------------------------------ |
| best    | (N\log N) (intuition pratique) |
| average | ?                              |
| worst   | (N^{3/2})                      |

---

**Important**

Shellsort :

* bien meilleur que (N^2),
* très utilisé en pratique,
* simple,
* peu de mémoire.

## 8. Shuffle Sort

## Mauvaise idée

Attribuer un random à chaque élément puis trier.

Correct mathématiquement mais inutilement lent.

---

**Mauvais shuffle**

```python
for i in range(N):
    r=random(0,N-1)
    swap(a[i],a[r])
```

**biaisé.**

---

**Knuth Shuffle / Fisher-Yates**

## Correct

```python
for i in range(N-1,0,-1):

    r=random(0,i)

    swap(a[i],a[r])
```

---

**Pourquoi ça marche ?**

À chaque étape :

* on fixe définitivement une position.

Chaque permutation devient équiprobable.

---

**Complexité**

Temps :

$O(N)$

Mémoire :

$O(1)$

## 9. Random generators & Seeds

## Seed

Valeur initiale du générateur pseudo-aléatoire.

Même seed :

**même suite pseudo-aléatoire.**

---

**Danger**

Seed prévisible :

* temps,
* millisecondes,
* timestamp

⇒ shuffle prévisible.

---

**Exemple poker**

Mauvais RNG :

* seulement quelques millions de shuffles possibles,
* exploitable.

---

**Solution**

Utiliser :

* CSPRNG,
* hardware RNG,
* Fisher-Yates unbiased.
