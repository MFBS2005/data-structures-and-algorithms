```python
#Exercice5:
class Stack:
  def __init__(self):
    self.data=[None]*10
    self.N=0
  def isEmpty(self):
    return self.N==0

  def push(self,item):
    if self.N==len(self.data):
      self.resize(len(self.data)*2)
    self.data[self.N]=item
    self.N+=1

  def pop(self):
    if self.N!=0:
      self.N-=1
      item=self.data[self.N]
      self.data[self.N]=None
      if self.N==len(self.data)//4:
        self.resize(len(self.data)//2)
      return item
    else:
      print("Stack is empty")
      #raise Exception("Stack is empty!")

  def resize(self,capacity):
    copy=[None]*capacity
    for i in range(self.N):
      copy[i]=self.data[i]
    self.data=copy
  def top(self):
    if self.N!=0:
      return self.data[self.N-1]
    else:
      print("Stack is empty")
  def usedSpace(self):
    return (self.N*100)/len(self.data)
  def size(self):
    return self.N



stack = Stack()

print("## testing push and pop ##")
print("")
stack.push("K")
stack.push("C")
stack.push("A")
stack.push("T")
stack.push("S")

print(stack.pop())
print(stack.pop())
print(stack.pop())
print(stack.pop())
print(stack.pop())

stack.push("A")
stack.push("B")
stack.push("C")

print("")
print("## testing top and empty ##")
print("")

print("Elements on this stack:", stack.size())
print("Is this stack empty?", stack.isEmpty())
print("Whtat is the TOP element:", stack.top())
print("Elements on this stack:", stack.size())
print("Is this stack empty?", stack.isEmpty())
print("")

stack.pop()
stack.pop()
stack.pop()

print("## testing resize ##")
print("")
for i in range(100):
  stack.push("A")
print("Used array space:", int(stack.usedSpace()), "%")
for i in range(1000):
  stack.push("A")
print("Used array space:", int(stack.usedSpace()), "%")
for i in range(800):
  stack.pop()
print("Used array space:", int(stack.usedSpace()), "%")


```

# Exercice 7:
This exercise computes Fibonacci numbers. The first cell (`fibIter`) returns the n-th Fibonacci number using an iterative approach (no recursion), keeping only the two previous values `a` and `b` at each step. The second cell builds the full list of Fibonacci numbers up to n, storing each value as it is computed.

```python
def fibIter(n):
    if n <= 0:
        return 0
    if n == 1:
        return 1

    a = 0
    b = 1
    i = 2

    while i <= n:
        aa = b
        bb = a + b
        a = aa
        b = bb
        i += 1
    return b
print(fibIter(10))
```
