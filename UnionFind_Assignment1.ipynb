```python
#Exercice1:
class quickfind:
  def __init__(self,n):
    self.id=[i for i in range(n)]
  def output(self,msg):
    while len(msg)!=7:#len of the longest word between before and after so that the printed resulut is symetric
      msg+=" "
    for i in range(len(self.id)):
      msg+=" "+str(i)+" "
    print(msg)
    print("      ",self.id)
    print("       ------------------------------")
  def find(self,p):
    self.nb+=1
    return self.id[p]
  def union(self,p,q):
    pid=self.find(p)
    qid=self.find(q)
    if not pid==qid:
      self.output("Before:")
      for i in range(len(self.id)):
        if self.id[i]==pid:
          self.id[i]=qid
          self.nb+=1
        self.nb+=1
      self.output("After:")
      print("The number of accesses of the array is: ",self.nb)
      print("")
    else:
      print("already connected")
  def unionlist(self,l):
    for i, (p,q) in enumerate(l):
      self.nb=0
      print("union",l[i],": ")
      self.union(p,q)



qf=quickfind(10)
l=[(9,0),(3,4),(5,8),(7,2),(2,1),(5,7),(0,3),(4,2)]
qf.unionlist(l)
```
